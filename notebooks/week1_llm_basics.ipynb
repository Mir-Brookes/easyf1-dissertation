```python
from dotenv import load_dotenv
import os
import requests

def ask_gemini(system_prompt, user_question):
    url = f"https://generativelanguage.googleapis.com/v1beta/models/gemini-2.5-flash-lite:generateContent?key={API_KEY}"

    payload = {
        "contents":[{
            "role": "user",
            "parts":[{"text": user_question}]
        }],

        "systemInstruction":{
            "parts":[{"text": system_prompt}]
        }
    }
    response = requests.post(url, json=payload, timeout=30)
    result = response.json()

    if "error" in result:
        return f"Error:{result['error']['message'][:150]}"
    else:
        return result["candidates"][0]["content"]["parts"][0]["text"]

print("Function Ready")
```

```python
system_prompt_1 = """
You are an aggressive F1 race strategist who always priortises attacking over 
defending. you favor bold and risky moves that could win or lose the race.
keep your answer to 3 sentence maximum """

question = "It's lap 32 at Silverstone. Hamilton is on 18-lap-old medium tyres. Safety car just deployed. What should Mercedes do?"

print('Prompt_1 - Aggressive:\n')
print(ask_gemini(system_prompt_1, question))
```

```python
system_prompt_2 = """
You are a conservative F1 race strategist who priortises finishing race above all else
You avoid risk at all cost and always choose the safest option.
Keep your answer to 3 sentence maximum """

print('Prompt_2 - Conservative:\n')
print(ask_gemini(system_prompt_2, question))
```

```python
system_prompt_3 = """
You are a F1 data scientist, you only speak in numbers, percentages and probabilites.
Never give an opinion - Only state what the data suggests.
Format your response as bullet points with numbers only"""

print('Prompt_3 - Data Scientist')
print(ask_gemini(system_prompt_3, question))
```

```python
system_prompt_4 ="""
You are explaining F1 strategy to someone who has never watched a 
race before. Use simple everyday language, no jargon, and use 
analogies to everyday life to explain what should happen and why."""

print("Prompt_4 - Simple Explainer")
print(ask_gemini(system_prompt_4, question))
```
